Indexación (Acceso a Elementos)

```python
frutas = ["manzana", "platano", "cereza"]

print(frutas[0])    # manzana  -> primer elemento
print(frutas[2])    # cereza   -> tercer elemento
print(frutas[-1])   # cereza   -> último elemento
print(frutas[-2])   # platano  -> penúltimo elemento
```

```python
# Preguntar cuantas calificaciones desea promediar , y promediarlas
def promedio (sumatoria , n):
    return sumatoria / n
n = int (input("¿Cuantas calificaciones quieres promediar?"))

sumatoria = 0
print( f"se van a pedir {n} calificaciones")
for c in range (1, n+1):
    calif =float(input (f"dame la calificacion {c} ") )
    sumatoria = sumatoria + calif

print(f" El promedio de {n} calificaciones es {promedio(sumatoria , n)}")
```

```python
filas = int(input("cantidad de filas"))
columnas = int(input("cantidad de columnas"))
matriz = []
for f in range (filas):
    fila_actual = []
    for c in range (columnas):
         Valor = int(input(f"Valor para [{f}] [{c}]"))
         fila_actual.append(Valor)
    matriz.append(fila_actual)
         
print("Impresion de matriz")
for f in range (filas):
     for c in range(columnas):
          print(matriz [f] [c] , end= " ")
     print()                   
```

```python
dimension = int(input("cantidad de dimensión"))
matriz = []
for f in range (dimension):
    fila_actual = []
    for c in range (dimension):
         Valor = 0
         if f == c:
              Valor = 1
         fila_actual.append(Valor)
    matriz.append(fila_actual)
        
print("Impresion de matriz")
for f in range (dimension):
     for c in range(dimension):
          print(matriz [f] [c] , end= "  ")
     print()
```
