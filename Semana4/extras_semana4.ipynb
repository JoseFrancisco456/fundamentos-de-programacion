Extra 1: Suma de todos los elementos de una matriz

```python
def imprimir_matriz(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

def suma_matriz(matriz):
    total = 0
    for fila in matriz:
        for elemento in fila:
            total = total + elemento
    return total
    

matriz = [
[1, 2, 3], 
[4, 5, 6], 
[7, 8, 9]]

imprimir_matriz(matriz)
resultado = suma_matriz(matriz)
print(f"La suma de todos los elementos de la matriz es: {resultado}")
```

Extra 2: Matriz con suma por fila


```python
def imprimir_matriz(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()
def suma_fila(fila):
    total = 0
    for elemento in fila:
        total = total + elemento
    return total

matriz = [
    [3, 1, 4], 
    [1, 5, 9], 
    [2, 6, 5]]

imprimir_matriz(matriz)

for f in range(len(matriz)):
    suma = suma_fila(matriz[f])
    print(f"Suma de la fila {f}: {suma}")
```

Extra 3: Producto de una columna sin usar el operador *

```python
def multiplicar_columna(matriz, columna):
    resultado = 1

    for fila in matriz:
        numero = fila[columna]

        suma = 0

        for i in range(resultado):
            suma = suma + numero

        resultado = suma

    return resultado


matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
    [2, 4, 6]
]

columna = int(input("Numero de columna: "))

if 0 <= columna <= 2:
    resultado = multiplicar_columna(matriz, columna)
    print(f"El producto de la columna {columna} es: {resultado}")
else:
    print("Error: la columna debe ser 0, 1 o 2.")
```

Extra 4: Tabla de Pitágoras de tamaño variable

```python
def generar_tabla(tamaño):
    tabla = []

    for fila in range(1, tamaño + 1):
        fila_actual = []

        for columna in range(1, tamaño + 1):
            resultado = 0

            for i in range(columna):
                resultado = resultado + fila

            fila_actual.append(resultado)

        tabla.append(fila_actual)

    return tabla


def imprimir_tabla(tabla):
    for fila_actual in tabla:
        for numero in fila_actual:
            print(numero, end="\t")
        print()


def consultar_producto(tabla, fila, columna):
    return tabla[fila - 1][columna - 1]


tamaño = int(input("Tamano de la tabla: "))

if 2 <= tamaño <= 5:

    tabla = generar_tabla(tamaño)

    imprimir_tabla(tabla)

    fila = int(input("Renglon (factor): "))
    columna = int(input("Columna (factor): "))

    if 1 <= fila <= tamaño and 1 <= columna <= tamaño:
        resultado = consultar_producto(tabla, fila, columna)
        print(f"El producto de {fila} x {columna} es: {resultado}")
    else:
        print("Error: el renglón y la columna deben estar dentro de la tabla.")

else:
    print("Error: el tamaño debe estar entre 2 y 5.")
```
