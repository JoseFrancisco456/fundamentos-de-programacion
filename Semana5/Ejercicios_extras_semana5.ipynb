Extra 1: Sistema de calificaciones con tuplas

```python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

nueva_calificacion1 = float(input("Nueva calificación 1: "))
nueva_calificacion2 = float(input("Nueva calificación 2: "))

nuevas_calificaciones = calificaciones + (nueva_calificacion1, nueva_calificacion2)

print("Tercera calificación:", nuevas_calificaciones[2])

lista = list(nuevas_calificaciones)

lista.sort(reverse=True)

def sumar_calificaciones(tupla):

    suma = 0

    for calificacion in tupla:

        suma += calificacion

    return suma

print("Nueva tupla:", nuevas_calificaciones)

print("Lista ordenada (mayor a menor):", lista)

print("Suma total:", sumar_calificaciones(nuevas_calificaciones))
```

Extra 2: Agenda de contactos con búsqueda

```python
agenda_contactos = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}

nombre = input("Nombre del nuevo contacto: ")

telefono = input("Teléfono del nuevo contacto: ")

agenda_contactos[nombre] = telefono

print("Contactos registrados:", ", ".join(agenda_contactos.keys()))

def buscar_telefono(agenda_contactos, nombre):

    return agenda_contactos.get(nombre, "Contacto no encontrado")

buscar = input("Nombre a buscar: ")

print(f"El teléfono de {buscar} es: {buscar_telefono(agenda_contactos, buscar)}")
```

Extra 3: Calculadora segura con manejo de excepciones

```python
try:
    num1 = int(input("Ingresa el primer número entero: "))
    num2 = int(input("Ingresa el segundo número entero: "))
    try:
        resultado = num1  / num2
        print (f"{num1} entre {num2} es : {resultado}")
    except ZeroDivisionError:
        print("Error: No es posible dividir entre cero. Ingresa un numero distino a 0")

except ValueError:
    print("Error: Debes ingresar números enteros válidos. Inténtalo de nuevo.")
```

Extra 4: Analizador de mensajes con strings

```python
mensaje = "Python es un lenguaje poderoso"

print("Longitud del mensaje:", len(mensaje))

print("En mayúsculas:", mensaje.upper())

print("Texto reemplazado:", mensaje.replace("Python", "programación"))

def contar_palabras(texto):

    return len(texto.split())

print("Palabras totales:", contar_palabras(mensaje))
```
