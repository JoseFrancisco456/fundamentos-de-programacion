Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario votos = {"Rojo": 0, "Azul": 0, "Verde": 0}. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con items() el color ganador junto con su total.



```python
candidatos = {
    "Rojo": 0,
    "Verde": 0,
    "Azul": 0
}

cantidad = int(input("¿Cuántos votos van a ser? "))

conteo = 1

while conteo <= cantidad:
    voto = input(f"Ingresa el voto {conteo} (Rojo, Verde o Azul): ")
    voto = voto.capitalize()
    if voto in candidatos:
        print(f"Voto para {voto}")
        candidatos[voto] += 1
        conteo += 1
    else:
        print("El voto no es válido")
ganador = ""
mayor = -1

for k, v in candidatos.items():
    print(f"Opción {k}, votos: {v}")
    if v > mayor:
        mayor = v
        ganador = k
print(f"Ganador: {ganador} con {mayor} votos")
```

Desafío Extra 3: Validación robusta de entrada
Escribe un programa que pida al usuario dos números enteros dentro de un bloque try-except. Si el usuario ingresa texto o un valor vacío, muestra un mensaje controlado y repite la solicitud con un ciclo while. Luego pide la división del primero entre el segundo, con un bloque específico para ZeroDivisionError.



```python
while True:
    try:
        num1 = int(input("Ingresa el primer número entero: "))
        num2 = int(input("Ingresa el segundo número entero: "))
        break  
    except ValueError:
        print("Error: Debes ingresar números enteros válidos. Inténtalo de nuevo.")


try:
    resultado = num1 / num2
    print(f"La división de {num1} entre {num2} es: {resultado}")
except ZeroDivisionError:
    print("Error: No se puede dividir entre cero.")
```

Desafío Extra 4: Analizador de textos
Crea un programa que reciba una frase, imprima su longitud con len(), la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo.

```python
mensaje = input("Ingresa una frase: ")

print("Longitud:", len(mensaje))

print("En mayúsculas:", mensaje.upper())

print("Reemplazo:", mensaje.replace("Python", "programación"))

def contar_palabras(texto):

    return len(texto.split())

print("Cantidad de palabras:", contar_palabras(mensaje))
```

Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo while y opciones numéricas que permita ejecutar de forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en el diccionario contactos, (3) dividir dos números con manejo de ZeroDivisionError y (4) analizar un mensaje de texto. La opción (5) debe finalizar el programa.

```python
def mostrar_tupla():

    numeros = (8, 3, 12, 5, 1)

    lista = list(numeros)

    lista.sort()

    print("Tupla ordenada:", lista)


def buscar_telefono():

    contactos = {
        "Ana": "555-0101",
        "Luis": "555-0102",
        "Mía": "555-0103"
    }

    nombre = input("Nombre a buscar: ")

    print("Teléfono:", contactos.get(nombre, "Contacto no encontrado"))


def dividir():

    try:

        primero = int(input("Primer número: "))
        segundo = int(input("Segundo número: "))

        print("Resultado:", primero / segundo)

    except ValueError:

        print("Error: Ingresa números enteros válidos.")

    except ZeroDivisionError:

        print("Error: No se puede dividir entre cero.")


def analizar_mensaje():

    mensaje = input("Ingresa un mensaje: ")

    print("Longitud:", len(mensaje))

    print("Mayúsculas:", mensaje.upper())

    print("Reemplazo:", mensaje.replace("Python", "programación"))

    print("Palabras:", len(mensaje.split()))


while True:

    print(" Mini Menu ")
    print("1. Mostrar tupla ordenada")
    print("2. Buscar teléfono")
    print("3. Dividir números")
    print("4. Analizar mensaje")
    print("5. Finalizar")

    try:

        opcion = int(input("Selecciona una opción: "))

        if opcion == 1:

            mostrar_tupla()

        elif opcion == 2:

            buscar_telefono()

        elif opcion == 3:

            dividir()

        elif opcion == 4:

            analizar_mensaje()

        elif opcion == 5:

            print("Programa finalizado.")

            break

        else:

            print("Opción no válida.")

    except ValueError:

        print("Ingresa una opción numérica.")
```
