Ejercicio 11: Suma de los primeros N números
Enunciado: Crea un programa que pida un número entero positivo N y calcule con while la suma de los números del 1 al N. Entrada: 5 Salida: La suma del 1 al 5 es: 15

```python
numero = int(input("Ingresa un número: "))
total = 0
x = 1

while x <= numero:
    total = total + x
    x = x + 1

print(f"La suma del 1 al {numero} es: {total}")
```
