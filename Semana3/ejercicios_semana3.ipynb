Ejercicio 1: Algoritmo con decisión para descuento por monto
Enunciado: Escribe en lenguaje natural los pasos del algoritmo (modelo Entrada-Proceso-Salida) para que una tienda aplique un 10% de descuento cuando la compra sea de $300 o más, y muestre el monto del descuento y el total a pagar. Entrada: monto = 500 Proceso: descuento = 500 * 0.10, total = 500 - 50 Salida: Monto: 500. Descuento: 50.0. Total a pagar: 450.0

```python
monto = int(input("Ingrese el monto: "))
descuento = 0.10
if monto >= 300 :
    montod = monto * descuento 
    total= monto-montod
    print("El monto es de :" , {monto})
    print("El total del descuento es:" , {montod})
    print("El total con descuento es:" , {total})
```

Ejercicio 2: Años bisiestos en un rango
Enunciado: Crea un programa que pida dos años (inicio y fin) y muestre en una línea todos los años bisiestos de ese rango. Regla: un año es bisiesto si es divisible entre 4 pero no entre 100, o si es divisible entre 400. Usa el for con range y operadores lógicos (and/or). Entrada: 2000 y 2010 Proceso: 2000 % 400 == 0 (bisiesto), 2004 % 4 == 0 y 2004 % 100 != 0 (bisiesto), 2008 (bisiesto) Salida: Años bisiestos: 2000, 2004, 2008

```python
año1 = int(input("Ingresa un año "))
año2 = int(input("Ingresa otro año "))
año=0
for año in range (año1, año2 + 1 ):
    if (año % 4 == 0 and año % 100 != 0) or (año % 400 == 0 ):
         print(f"es un año bisiesto {año}")
    else: 
         print(f"no es bisiesto {año}")
```

Ejercicio 3: Contador de dígitos
Enunciado: Crea un programa que pida un número entero positivo y cuántos dígitos tiene, usando while con el operador división entera // (cada iteración elimina el último dígito). Entrada: 4572 Proceso: 4572 -> 457 -> 45 -> 4 -> 0 (4 eliminaciones) Salida: El número 4572 tiene 4 dígitos

```python
num = int(input("numero: "))
original_num = num
contador= 0

if num ==0:
    contador==0
else:
    while num > 0:
        num //= 10
        contador += 1
    num = original_num
print(f"el numero {num} tiene {contador} digitos")
```

Entrada: 1234 Proceso: 4, luego 43, luego 432, luego 4321 Salida: El número invertido de 1234 es 4321

```python
num = int (input("Ingresa un numero entero positivo: "))
invertido= 0
while num != 0:
    r= num % 10
    invertido = invertido * 10 + r
    num = num // 10
print(f"{invertido}")
```

Enunciado: Sin usar los operadores / ni //, calcula el cociente y el residuo de una división entre enteros positivos restando el divisor al dividendo con un while; el cociente es la cantidad de restas realizadas. Entrada: 17 y 5 Proceso: 17-5=12, 12-5=7, 7-5=2 (3 restas, ya no se puede restar más) Salida: 17 ÷ 5 = 3 con residuo 2

```python
dividendo = int(input("Dame un numero mayor a el divisor"))
divisor = int (input(" Ingresa un divisor no pongas 0 "))

if divisor <= 0:
    print ("El numero debe ser mayor a 0")
else:
    cociente = 0
    residuo = dividendo

    while residuo >= divisor:
        residuo= residuo - divisor
        cociente = cociente + 1
print(f"{dividendo} / {divisor} = {cociente} con residuo {residuo}")
print(f"Termina el proceso")
```

Enunciado: Crea un programa que pida un entero positivo n y calcule la suma alternada 1 - 2 + 3 - 4 + ... hasta n. Usa un for y decide el signo de cada término con el módulo %. Entrada: 6 Proceso: 1 - 2 + 3 - 4 + 5 - 6 = -3 Salida: El resultado de la serie es: -3

```python
n= int (input("ingresa un numero entero: "))
suma= 1
for i in range (1+ n +1):
    if i % 2 ==0:
        suma += i
    else: suma -= i
print(f"El resultado de la serie es : {suma}")
```

Ejercicio 12: Contar números pares
Enunciado: Crea un programa que cuente cuántos números pares se ingresan, y se detenga cuando el usuario ingrese un número negativo. Entrada: 2, 5, 8, 10, -1 Proceso: pares detectados: 2, 8, 10 Salida: Cantidad de números pares: 3

```python
pares = 0

numero = int(input("Ingresa un número (negativo para terminar): "))

while numero >= 0:
    if numero % 2 == 0:
        pares = pares + 1

    numero = int(input("Ingresa un número (negativo para terminar): "))

print("Cantidad de números pares:", pares)
```
