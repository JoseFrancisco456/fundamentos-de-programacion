Extra 1: División de cuenta con propina

```python
total_cuenta = float(input("Total de la cuenta :"))
porcentaje_propina = float(input("Porcentaje de propina :"))
personas = int(input("Numero de personas :"))

propina = porcentaje_propina / 100 * total_cuenta
total_con_propina = total_cuenta + propina
pago_persona = total_con_propina / personas
print(f"Propina : ${propina:.2f}")
print(f"total con propina : ${total_con_propina:.2f}")
print(f"Pago por persona : ${pago_persona:.2f}") 
```

Extra 2: Conversor de minutos a días, horas y minutos

```python
total_minutos = int(input("Total de minutos: "))

dias = total_minutos // 1440
minutos_restantes = total_minutos % 1440
horas = minutos_restantes // 60
minutos = minutos_restantes % 60

print(f"{total_minutos} minutos = {dias} día(s), {horas} hora(s), {minutos} minuto(s)")
```

Extra 3: Calificación final ponderada

```python
parcial_1 = float(input("Parcial 1 (30%): "))
parcial_2 = float(input("Parcial 2 (30%): "))
parcial_3 = float(input("Parcial 3 (40%): "))

calificacion_final = (  parcial_1 * 0.30 + parcial_2 * 0.30 + parcial_3 * 0.40)

print(f"Tu calificación final es: {calificacion_final:.2f}")
```

Extra 4: Conversor de moneda (MXN a USD y EUR)

```python
cantidad_mxn = float(input("Cantidad en MXN: "))
tipo_cambio_usd = float(input("Tipo de cambio USD: "))
tipo_cambio_eur = float(input("Tipo de cambio EUR: "))

usd = cantidad_mxn / tipo_cambio_usd
eur = cantidad_mxn / tipo_cambio_eur

print(f"${cantidad_mxn:.2f} MXN equivalen a:")
print(f"USD: {usd:.2f}")
print(f"EUR: {eur:.2f}")
```
